```python
df = pd.read_csv(data)
```

```python
df.isnull().sum()
```

```python
df[['horsepower', 'acceleration']] = df[['horsepower', 'acceleration']].fillna(0.0)
```

```python
from sklearn.model_selection import train_test_split
```

```python
df_full_train, df_test = train_test_split(df, test_size=0.2, random_state=1)
df_train, df_val = train_test_split(
    df_full_train, test_size=0.25, random_state=1
)
```

# Q1 - Original splitting feature

```python
from sklearn.feature_extraction import DictVectorizer
```

```python
y_train = df_train['fuel_efficiency_mpg'].values
del df_train['fuel_efficiency_mpg']
```

```python
train_dicts = df_train.fillna(0).to_dict(orient='records')

dv = DictVectorizer(sparse=False)
X_train = dv.fit_transform(train_dicts)
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.tree import DecisionTreeRegressor
from sklearn.feature_extraction import DictVectorizer
from sklearn.metrics import roc_auc_score
```

```python
dt = DecisionTreeRegressor(max_depth=1)
dt.fit(X_train, y_train)
```

```python
from sklearn.tree import export_text

print(export_text(dt, feature_names=list(dv.get_feature_names_out())))
```

# Q2 - RMSE on validation

```python
from sklearn.ensemble import RandomForestRegressor
```

```python
rf = RandomForestRegressor(n_estimators = 10, random_state=1, n_jobs=-1)
rf.fit(X_train, y_train)
```

```python
y_val = df_val['fuel_efficiency_mpg'].values
del df_val['fuel_efficiency_mpg']
```

```python
val_dicts = df_val.fillna(0).to_dict(orient='records')
X_val = dv.transform(val_dicts)
```

```python
y_pred = rf.predict(X_val)
```

```python
from sklearn.metrics import root_mean_squared_error

rmse = root_mean_squared_error(y_val, y_pred)
print(round(rmse, 3))
```

# Q3 - Number of estimators

```python
scores = []

for n in [10, 50, 100, 150]:
    rf = RandomForestRegressor(n_estimators = n, random_state=1, n_jobs=-1)
    rf.fit(X_train, y_train)
    y_pred = rf.predict(X_val)

    rmse = root_mean_squared_error(y_val, y_pred)
    scores.append((n, round(rmse, 3)))
```

```python
scores
```

# Q4 - Best max depth

```python
scores = []

for d in [10, 15, 20, 25]:
    for n in [10, 50, 100, 150]:
        rf = RandomForestRegressor(n_estimators = n,
                                   max_depth = d,
                                   random_state=1,
                                   n_jobs=-1)
        rf.fit(X_train, y_train)
        y_pred = rf.predict(X_val)
    
        rmse = root_mean_squared_error(y_val, y_pred)
        scores.append((d, n, round(rmse, 3)))
```

```python
columns = ['max_depth', 'n_estimators', 'rmse']
df_scores = pd.DataFrame(scores, columns=columns)
df_scores
```

```python
df_scores.groupby('max_depth')['rmse'].mean()
```

# Q5 - Most important feature

```python
rf = RandomForestRegressor(n_estimators = 10, max_depth = 20, random_state=1, n_jobs=-1)
rf.fit(X_train, y_train)
```

```python
importances = pd.DataFrame({
    'feature': dv.get_feature_names_out(),
    'importance': rf.feature_importances_
})
importances
```

```python
importances.sort_values('importance', ascending=False)
```

# Q6

```python
import xgboost as xgb
```

```python
features = list(dv.get_feature_names_out())
dtrain = xgb.DMatrix(X_train, label=y_train, feature_names=features)
```

```python
dval = xgb.DMatrix(X_val, label=y_val, feature_names=features)
```

```python
xgb_params = {
    'eta': 0.3, 
    'max_depth': 6,
    'min_child_weight': 1,
    
    'objective': 'reg:squarederror',
    'nthread': 8,
    
    'seed': 1,
    'verbosity': 1,
}

model = xgb.train(xgb_params, dtrain, num_boost_round=10)
```

```python
y_pred = model.predict(dval)
```

```python
rmse = root_mean_squared_error(y_val, y_pred)
```
