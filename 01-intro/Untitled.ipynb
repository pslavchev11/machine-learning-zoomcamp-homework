```python
df_asia = df[df.origin == 'Asia']
```

```python
two_col = df_asia[['vehicle_weight', 'model_year']]
```

```python
first_7 = two_col.head(7)
```

```python
X = first_7.to_numpy()
```

```python
XTX = X.T.dot(X)
```

```python
XTX_inv = np.linalg.inv(XTX)
```

```python
XTX_inv
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
w = (XTX_inv.dot(X.T)).dot(y)
```
